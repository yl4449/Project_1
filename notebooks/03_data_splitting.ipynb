```python
import pandas as pd
```

```python
data = pd.read_csv("../data/cleaned_with_missing_data.csv")

data.head()
```

```python
# Check missing values
missing_values = data.isnull().sum()

print("Missing values by column:")
print(missing_values)
```

```python
# Split by missing vs non-missing animal_id
missing_animal_id = data[data["animal_id"].isna()].copy()
no_missing_animal_id = data[data["animal_id"].notna()].copy()

print("Missing animal_id rows:", len(missing_animal_id))
print("No missing animal_id rows:", len(no_missing_animal_id))
```

```python
# Keep rows where animal_id is not missing
no_missing_animal_id = data[data["animal_id"].notna()].copy()

# Save to a new CSV file
no_missing_animal_id.to_csv("../data/no_missing_animal_id.csv", index=False)

print("Saved no_missing_animal_id to ../data/no_missing_animal_id.csv")
```
